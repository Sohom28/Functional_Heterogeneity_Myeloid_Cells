```python
import pandas as pd

def separate_overlapping_indices(*dfs):
    """
    Separates overlapping indices from multiple DataFrames.
    Returns:
        - Five DataFrames with unique indices.
        - One DataFrame (`overlapped_index_df`) containing all unique overlapping indices.
    """
    # Step 1: Get all indices from the DataFrames
    all_indices = [set(df.index) for df in dfs]
    
    # Step 2: Identify unique and overlapping indices
    all_indices_combined = set.union(*all_indices)
    index_counts = {}
    
    for idx in all_indices_combined:
        count = sum(idx in df_index for df_index in all_indices)
        index_counts[idx] = count

    overlapping_indices = {idx for idx, count in index_counts.items() if count > 1}
    
    # Step 3: Create cleaned DataFrames without overlapping indices
    cleaned_dfs = [df.loc[~df.index.isin(overlapping_indices)] for df in dfs]
    
    # Step 4: Create a single DataFrame for unique overlapping indices
    overlapping_data = {idx: [] for idx in overlapping_indices}
    
    for df in dfs:
        for idx in df.index:
            if idx in overlapping_indices:
                overlapping_data[idx].append(df.loc[idx])
    
    # Convert overlapping data dictionary into a DataFrame
    overlapped_index_df = pd.DataFrame.from_dict(overlapping_data, orient="index")

    return cleaned_dfs + [overlapped_index_df]

# Example DataFrames
#filtered_new_archetype_1_df = pd.DataFrame({"A": [1, 2, 3]}, index=["a", "b", "c"])
#filtered_new_archetype_2_df = pd.DataFrame({"B": [4, 5, 6]}, index=["b", "c", "d"])
#filtered_new_archetype_3_df = pd.DataFrame({"C": [7, 8, 9]}, index=["c", "d", "e"])
#filtered_new_archetype_4_df = pd.DataFrame({"D": [10, 11, 12]}, index=["d", "e", "f"])
#filtered_new_archetype_5_df = pd.DataFrame({"E": [13, 14, 15]}, index=["e", "f", "g"])

# Apply function
cleaned_dfs = separate_overlapping_indices(filtered_new_archetype_1_df, 
                                           filtered_new_archetype_2_df, 
                                           filtered_new_archetype_3_df, 
                                           filtered_new_archetype_4_df, 
                                           filtered_new_archetype_5_df)

# Unpacking results
(filtered_new_archetype_1_cleaned_df, 
 filtered_new_archetype_2_cleaned_df, 
 filtered_new_archetype_3_cleaned_df, 
 filtered_new_archetype_4_cleaned_df, 
 filtered_new_archetype_5_cleaned_df, 
 overlapped_index_df) = cleaned_dfs

# Display results
print("Filtered Archetype 1 (Cleaned):\n", filtered_new_archetype_1_cleaned_df)
print("Filtered Archetype 2 (Cleaned):\n", filtered_new_archetype_2_cleaned_df)
print("Filtered Archetype 3 (Cleaned):\n", filtered_new_archetype_3_cleaned_df)
print("Filtered Archetype 4 (Cleaned):\n", filtered_new_archetype_4_cleaned_df)
print("Filtered Archetype 5 (Cleaned):\n", filtered_new_archetype_5_cleaned_df)
print("Overlapping Indices DF:\n", overlapped_index_df)
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd

color_map = {
    "Normal": "#0000FF",      # Blue
    "Stage_1": "#1E90FF",     # DodgerBlue
    "Stage_2": "#228B22",     # ForestGreen
    "Stage_3": "#FF4500",     # OrangeRed
    "Stage_4": "#800000"      # Maroon
}


# Get unique samples (stages) from the DataFrame
sample_unique = stacked_bar_df['Sample'].unique()

# Group the data by 'archetype_no' and 'Sample' to calculate counts
counts = stacked_bar_df.groupby(['archetype_no', 'Sample']).size().unstack(fill_value=0)

# Calculate percentage
counts_percentage = counts.div(counts.sum(axis=1), axis=0) * 100

# Plot the stacked bar chart for percentages with custom colors
ax = counts_percentage.plot(kind='bar', stacked=True, figsize=(12, 8),
                             color=[color_map[sample] for sample in counts_percentage.columns])

# Add percentage labels on top of each section of the stacked bar
for i, bars in enumerate(ax.containers):
    # Add labels only if the percentage is greater than a threshold, e.g., 1%
    ax.bar_label(bars, label_type='center', fmt='%.2f%%', padding=3)

# Add labels and title
plt.xlabel('Archetype Number in Myeloid Cells')
plt.ylabel('Percentage of samples (%)')
plt.title('Percentage Stacked Bar Plot of Samples by Archetype Number in Myeloid Cells')

# Move the legend outside of the plot
plt.legend(title='Sample', bbox_to_anchor=(1.05, 1), loc='upper left', labels=color_map.keys())

# Adjust layout to make space for the legend
plt.tight_layout(rect=[0, 0, 0.85, 1])

directory = '/home/sohom/sohom/new_project/all_myeloid(5_archetypes)/Stacked bar and clustermap(Myeloid_Cells)'
filename = 'new_stacked_bar_plot_percentage_Myeloid_Cells.pdf'
# Save the plot as a PDF
# Save the plot to the specified directory
os.makedirs(directory, exist_ok=True)

file_path = os.path.join(directory, filename)
plt.savefig(file_path, format='pdf', dpi=300)

# Show the plot
plt.show()
```

```python
import os
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Define custom archetype color palette
custom_archetype_palette = {
    "Generalist": "#A52A2A",  # Brown
    "Specialist-1": "#FFA500",  # Orange
    "Specialist-2": "#808080",  # Gray
    "Specialist-3": "#008000",  # Green
    "Specialist-4": "#FFFF00",  # Yellow
    "Specialist-5": "#0000FF"   # Blue
}

# Map the colors to the archetypes in the DataFrame
col_colors = archetype_heatmap_df.set_index('index')['archetype_no'].map(custom_archetype_palette)

# Generate the clustermap
g = sns.clustermap(
    data=filtered_heatmap_df,
    z_score=0,  # Normalize rows
    row_cluster=True,  # Cluster rows
    col_cluster=False,  # Do not cluster columns
    row_colors=None,  # No colors for rows
    col_colors=col_colors,  # Apply column colors
    vmin=-3,  # Min value for heatmap scaling
    vmax=3,  # Max value for heatmap scaling
    cmap="coolwarm",  # Heatmap color scheme
    metric='euclidean',  # Distance metric for clustering
    figsize=(50, 35),  # Figure size
    cbar_pos=(-0.1, 0.2, 0.03, 0.6)  # Colorbar position on the left
)

# Hide row and column dendrograms
g.ax_row_dendrogram.set_visible(False)
g.ax_col_dendrogram.set_visible(False)

# Create legend handles for the archetypes
handles = [
    mpatches.Patch(color=color, label=archetype)
    for archetype, color in custom_archetype_palette.items()
]

# Add a legend for archetype colors
g.ax_heatmap.legend(
    handles=handles,
    title='Archetype Colors',
    bbox_to_anchor=(1.1, 1.02),  # Position legend outside heatmap
    loc='upper left',
    borderaxespad=0.,
    fontsize=25,
    title_fontsize=25
)

# Define the directory and file path for saving the plot
directory = '/home/sohom/sohom/new_project/all_myeloid(5_archetypes)/Stacked bar and clustermap(Myeloid_Cells)'
filename = 'clustermap_for_myeloid_archetypes.png'
file_path = os.path.join(directory, filename)

# Ensure the directory exists
os.makedirs(directory, exist_ok=True)

# Save the clustermap as a PNG file
g.savefig(file_path, format='png', dpi=500, bbox_inches='tight')

# Display the plot
plt.show()
```
